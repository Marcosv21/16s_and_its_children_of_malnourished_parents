## Renamed

```python
import os
import pandas as pd

# 1. Set your folder and CSV file paths
folder = '/home/marcos/colaboração/Pedro Melo/RICHARDT-GAMA-LANDGRAF_results/RICHARDT-GAMA-LANDGRAF_results/fastqs/16s'
csv_file = '/home/marcos/colaboração/Pedro Melo/RICHARDT-GAMA-LANDGRAF_results/RICHARDT-GAMA-LANDGRAF_results/meta_date_16s.csv'

# 2. Read the CSV file 
try:
    df = pd.read_csv(csv_file, delim_whitespace=True)
except:
    df = pd.read_csv(csv_file, sep=',')

# 3. Loop through the rows and rename the files
for index, row in df.iterrows():
    old_name = str(row['Name']).strip()
    new_name_base = str(row['sample_title']).strip()
    
    # Make sure the new name ends with .fastq
    if not new_name_base.endswith('.fastq'):
        new_name = f"{new_name_base}.fastq"
    else:
        new_name = new_name_base
        
    old_path = os.path.join(folder, old_name)
    new_path = os.path.join(folder, new_name)
    
    # Rename the file if it exists in the folder
    if os.path.exists(old_path):
        os.rename(old_path, new_path)
        print(f"[SUCCESS] Renamed: {old_name} -> {new_name}")
    else:
        print(f"[ERROR] File not found: {old_name}")

print("Done!")
```
